```python

from pathlib import Path

import tree_sitter_python as tspython
from tree_sitter import Language, Parser
```

```python
def main():

    folder_path = Path(r"C:\Users\gargd\Code_Assistant\python_files").expanduser()

    if not folder_path.exists():
        print("Error: Path does not exist.")
        return

    # Support both a single Python file and a folder
    if folder_path.is_file():
        python_files = [folder_path] if folder_path.suffix == ".py" else []
    else:
        python_files = sorted(folder_path.rglob("*.py"))

    if not python_files:
        print("No Python files found.")
        return

    print(f"\nFound {len(python_files)} Python file(s).")


if __name__ == "__main__":
    main()
```

```python
# 1. Initialize Tree-sitter Python parser
PYTHON_LANGUAGE = Language(tspython.language())
parser = Parser(PYTHON_LANGUAGE)


# 2. Function to visualize the syntax tree
def display_tree(node, source_code, level=0):

    indent = "  " * level

    # Get node's position
    start = node.start_point
    end = node.end_point

    # Display node information
    print(
        f"{indent}├── {node.type} "
        f"[{start.row + 1}:{start.column} - "
        f"{end.row + 1}:{end.column}]"
    )

    # Mark syntax errors, if any
    if node.is_error:
        print(f"{indent}    ERROR")

    if node.is_missing:
        print(f"{indent}    MISSING NODE")

    # Recursively visit child nodes
    for child in node.children:
        display_tree(child, source_code, level + 1)


# 3. Function to parse a Python file
def parse_file(file_path):

    print("\n" + "=" * 70)
    print(f"FILE: {file_path.name}")
    print("=" * 70)

    # Read source code
    source_code = file_path.read_bytes()

    # Generate syntax tree
    tree = parser.parse(source_code)

    # Get root node
    root_node = tree.root_node

    print("\nSYNTAX TREE:\n")

    # Visualize complete tree
    display_tree(root_node, source_code)

    print("\nParsing completed.")
    print("Contains syntax errors:", root_node.has_error)


# 4. Main function: accept a file or folder path
def main():

    folder_path = Path(r"C:\Users\gargd\Code_Assistant\python_files").expanduser()

    if not folder_path.exists():
        print("Error: Path does not exist.")
        return

    # Support both a single Python file and a folder
    if folder_path.is_file():
        python_files = [folder_path] if folder_path.suffix == ".py" else []
    else:
        python_files = sorted(folder_path.rglob("*.py"))

    if not python_files:
        print("No Python files found.")
        return

    print(f"\nFound {len(python_files)} Python file(s).")

    # Parse each file independently
    for file_path in python_files:
        try:
            parse_file(file_path)
        except (OSError, UnicodeError) as error:
            print(f"Could not read {file_path}: {error}")


if __name__ == "__main__":
    main()
```
